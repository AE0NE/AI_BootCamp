### 에이전트 구축 2

````python
from langchain.tools import tool

@tool
def add(a: int, b: int) -> int:
    """`a`와 `b` 덧셈.

    Args:
        a: First int
        b: Second int
    """
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """`a`와 `b` 곱셈.

    Args:
        a: First int
        b: Second int
    """
    return a * b

@tool
def divide(a: int, b: int) -> float:
    """`a`와 `b` 나눗셈.

    Args:
        a: First int
        b: Second int
    """
    return a / b
````

**요구사항 2 – 구조화된 분석 결과 정의**

Pydantic의 `BaseModel`을 사용하여 컴플레인 분석 결과의 구조를 정의하세요.

분석 결과에는 다음 필드가 포함되어야 합니다.

| 필드                | 설명                      |
| ----------------- | ----------------------- |
| `intent`          | 고객이 컴플레인을 접수한 핵심 의도     |
| `emotion`         | 고객의 주요 감정 상태            |
| `summary`         | 컴플레인 내용을 1~2문장으로 요약한 결과 |
| `next_action`     | 고객센터의 다음 행동             |
| `priority`        | 처리 우선순위                 |
| `reason`          | 분석 결과를 그렇게 판단한 이유       |
| `suggested_reply` | 고객에게 보낼 짧은 답변 초안        |

`next_action`은 반드시 다음 값 중 하나여야 합니다.

- 회신, 확인, 무시

`priority`는 반드시 다음 값 중 하나여야 합니다.

- 높음, 보통, 낮음

**요구사항 3 – 컴플레인 조회 Tool 구현**

사용자가 전달한 `complaint_id`와 일치하는 컴플레인 데이터를 조회하는 LangChain Tool을 구현하세요.

Tool 함수의 이름은 `get_complaint_by_id`로 작성하세요.

Tool은 다음 조건을 만족해야 합니다.

* 정수형 `complaint_id`를 입력받습니다.
* DataFrame의 `complaint_id` 컬럼에서 일치하는 데이터를 조회합니다.
* 조회 결과가 존재하면 첫 번째 행을 선택합니다.
* 결측값은 빈 문자열로 처리합니다.
* 조회 결과를 JSON 문자열로 변환하여 반환합니다.
* 한글이 유니코드 코드로 변환되지 않도록 처리합니다.
* 해당 ID가 없으면 `NOT_FOUND` 오류 정보를 JSON 문자열로 반환합니다.

**요구사항 4 – 컴플레인 분석 Agent 생성**

다음 기능을 수행하는 LangChain Agent를 생성하세요.

1. 사용자가 `complaint_id`를 전달하면 `get_complaint_by_id` Tool을 호출합니다.
2. Tool을 통해 조회한 컴플레인 내용만 사용하여 분석합니다.
3. 조회되지 않은 정보는 임의로 추측하지 않습니다.
4. 분석 결과는 요구사항 2에서 정의한 Pydantic 모델의 구조로 반환합니다.

Agent는 다음 기준에 따라 `next_action`을 판단해야 합니다.

* `회신`: 고객에게 안내, 사과 또는 추가 정보 요청이 필요한 경우
* `확인`: 주문, 배송, 환불, 재고, 제품 상태 또는 AS 가능 여부를 내부적으로 확인해야 하는 경우
* `무시`: 구매 사실이나 구체적인 요청이 없고 실제로 처리할 내용이 없는 경우

Agent는 다음 기준에 따라 `priority`를 판단해야 합니다.

* `높음`: 제품 파손, 안전 문제, 부작용, 환불 지연, 즉시 사용 불가
* `보통`: 배송 지연, 구성품 누락, 일반 교환·반품·AS 문의
* `낮음`: 처리할 수 없는 비구매성 비난 또는 정보가 거의 없는 경우

````python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-5.4-mini")
````

````python
response = model.invoke("랭스미스가 무엇인지 한 문장으로 설명해줘.")
print(f"답변: {response.content}")
````
