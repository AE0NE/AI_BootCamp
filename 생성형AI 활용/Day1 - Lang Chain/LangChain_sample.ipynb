````python
# 1. 모델 생성
model = init_chat_model("gpt-5.4-mini")

# 2. 도구 생성
@tool
def add(a: float, b: float) -> float:
    """두 숫자를 더합니다."""
    return a + b

@tool
def subtract(a: float, b: float) -> float:
    """첫 번째 숫자에서 두 번째 숫자를 뺍니다."""
    return a - b

# 3. 에이전트 생성
agent = create_agent(
    model,
    tools=[add, subtract]
)

# 4. 에이전트 실행
result = agent.invoke(
    {"messages": [
        {"role": "user", "content": "274에 356을 더하고 187을 빼면 뭐야?"}
    ]}
)

# 5. 최종 응답
result["messages"][-1].content
````

````python
import datetime

# UTC 기준 시간
utc_now = datetime.datetime.now(datetime.timezone.utc)

# 한국 시간으로 변환 (UTC + 9시간)
kst_timezone = datetime.timezone(datetime.timedelta(hours=9))
kst_now = utc_now.astimezone(kst_timezone)

print(f"현재 한국 시간: {kst_now.strftime('%Y-%m-%d %H:%M')}")
````

````python
@tool
def get_current_korea_time() -> str :
  "한국의 현재 날짜와 시간을 반환합니다."
  # UTC 기준 시간
  utc_now = datetime.datetime.now(datetime.timezone.utc)

  # 한국 시간으로 변환 (UTC + 9시간)
  kst_timezone = datetime.timezone(datetime.timedelta(hours=9))
  kst_now = utc_now.astimezone(kst_timezone)

  return f"현재 한국 시간: {kst_now.strftime('%Y-%m-%d %H:%M')}"
````

````python
agent = create_agent(model, tools=[get_current_korea_time])
````

````python
result = agent.invoke(
    {"messages" : [{"role" : "user", "content" : "현재 한국은 몇월 며칠 몇 시 몇분이야?"}]}
)
````

#### 에이전트 구축 실습 2

*   Franfurter Exchange rates data API [링크](https://frankfurter.dev/)

````python
import requests
from langchain.tools import tool

@tool
def get_exchange_rate(base: str = "USD", quote: str = "KRW", amount: float = 1.0) -> str:
    """
    특정 통화의 환율을 조회합니다.
    예: base='USD', quote='KRW'이면 1달러가 몇 원인지 조회합니다.
    amount를 넣으면 해당 금액을 환산합니다.
    """

    base = base.upper()
    quote = quote.upper()

    url = f"https://api.frankfurter.dev/v2/rate/{base}/{quote}"

    response = requests.get(url, timeout=10)
    response.raise_for_status()

    data = response.json()

    rate = data["rate"]
    date = data.get("date", "알 수 없음")

    converted = amount * rate

    return (
        f"기준일: {date}\n"
        f"환율: 1 {base} = {rate:,.2f} {quote}\n"
        f"환산: {amount:,.2f} {base} = {converted:,.2f} {quote}"
    )
````

````python
agent = create_agent(
    model="gpt-5.4-mini",
    tools=[get_exchange_rate],
    system_prompt="""
너는 환율 정보를 알려주는 친절한 금융 정보 assistant다.

사용자가 원-달러 환율, 달러 환율, USD/KRW 환율을 물어보면
반드시 get_exchange_rate Tool을 사용해 최신 환율을 조회해라.

한국에서 말하는 '원-달러 환율'은 보통
'1달러가 몇 원인지', 즉 USD -> KRW 환율을 의미한다.

답변에는 반드시 다음을 포함해라.
1. 기준일
2. Frankfurter API 기준이라는 설명

투자 조언처럼 말하지 말고, 조회된 환율 정보만 간단히 설명해라.
""",
)
````

````python
result = agent.invoke({ "messages": [{ "role": "user","content": "오늘 원-달러 환율 알려줘"}]})
````

````python
result
````

#### 에이전트 구축 실습3

앞서 구현한 `get_weather` Tool을 활용하여 다음 조건을 만족하는 날씨 안내 Agent를 구축하세요.

**실습 요구사항**

1. 사용자가 **한국의 지역 날씨**를 질문하면 `get_weather` Tool을 호출하여 고정된 날씨 정보(**-2도, 미세먼지 매우 나쁨**)를 제공하세요.
2. 사용자가 **한국 외 지역의 날씨**를 질문하면 Tool을 호출하지 않고, **“한국 외의 날씨 정보는 제공하지 않습니다.“**라는 메시지를 반환하세요.
3. `system_prompt`를 활용하여 Agent가 위 조건에 따라 동작하도록 설정하세요.

**실행 예시**
- 입력: “서울 날씨 알려줘” → `get_weather` 호출 후 날씨 정보 제공
- 입력: “도쿄 날씨 알려줘” → “한국 외의 날씨 정보는 제공하지 않습니다.”

````python
from langchain.chat_models import init_chat_model
from langchain.tools import tool
from langchain.agents import create_agent

# 1. 모델 생성
model = init_chat_model("gpt-5.4-mini")


# 2. 날씨 도구 생성
@tool
def get_weather(location: str) -> str:
    """한국의 날씨 정보를 제공합니다."""
    return f"{location}의 날씨는 눈이 오고, -2도, 미세먼지 나쁨입니다."


# 3. 에이전트 생성
agent = create_agent(
    model,
    tools=[get_weather],
    system_prompt="""
    당신은 한국의 날씨 정보를 제공하는 AI 에이전트입니다.

    다음 규칙을 반드시 지키세요.

    1. 사용자가 한국의 지역 날씨를 질문하면
       반드시 get_weather 도구를 호출하여 답변하세요.

    2. 사용자가 한국 외의 지역 날씨를 질문하면
       get_weather 도구를 호출하지 마세요.

    3. 한국 외의 지역에 대해서는 반드시 다음 문장으로만 답변하세요.
       "한국 외의 날씨 정보는 제공하지 않습니다."

    4. 날씨 정보를 임의로 만들어내지 마세요.
    """
)
````

````python
result = agent.invoke(
    {"messages": [
        {"role": "user", "content": "지금 서울 날씨 어때?"}
    ]}
)

result["messages"][-1].content
````

````python
result = agent.invoke(
    {"messages": [
        {"role": "user", "content": "지금 도쿄 날씨 어때?"}
    ]}
)

result["messages"][-1].content
````

````python
# 1) CSV 로드
import pandas as pd

CSV_PATH = "complaints_input.csv"

df = pd.read_csv(CSV_PATH)
````

````python
df.head()
````

````python
# 사본 생성
COMPLAINT_DF = df.copy()
````

````python
# 2) 스키마 정의

from typing import Literal
from pydantic import BaseModel, Field

class ComplaintAnalysis(BaseModel):
    intent: str = Field(description="고객이 이메일을 보낸 핵심 의도")
    emotion: str = Field(description="고객의 감정 상태. 예: 화남, 실망, 불안, 답답함, 중립")
    summary: str = Field(description="컴플레인 내용 1~2문장 요약")
    next_action: Literal["회신", "확인", "무시"] = Field(
        description="고객센터의 다음 행동"
    )
    priority: Literal["높음", "보통", "낮음"] = Field(
        description="처리 우선순위"
    )
    reason: str = Field(description="intent, next_action, priority를 그렇게 판단한 이유")
    suggested_reply: str = Field(description="고객에게 보낼 짧은 답변 초안")
````

````python
# 3) Tool 정의
import json
from langchain.tools import tool

@tool
def get_complaint_by_id(complaint_id: int) -> str:
    """
    complaint_id에 해당하는 고객 컴플레인 데이터를 CSV에서 조회합니다.
    반환값은 Agent가 읽기 쉬운 JSON 문자열입니다.
    """
    matched = COMPLAINT_DF[COMPLAINT_DF["complaint_id"] == complaint_id]

    if matched.empty:
        return json.dumps(
            {
                "error": "NOT_FOUND",
                "message": f"complaint_id={complaint_id} 데이터를 찾을 수 없습니다.",
            },
            ensure_ascii=False,
        )

    row = matched.iloc[0].fillna("").to_dict()

    return json.dumps(row, ensure_ascii=False)
````

````python
# 4) Agent 생성

agent = create_agent(
    model="gpt-5.4-mini",
    tools=[get_complaint_by_id],
    response_format=ComplaintAnalysis,
    system_prompt="""
너는 고객센터 컴플레인 분류 Agent다.

반드시 get_complaint_by_id Tool을 호출해서 고객 컴플레인 데이터를 조회한 뒤 분석해라.
사용자가 complaint_id를 주면, 그 ID의 데이터만 Tool로 조회해서 판단해라.
CSV에 없는 정보는 추측하지 마라.

분석 필드:
- intent: 고객의 핵심 의도
- emotion: 고객 감정 상태
- summary: 1~2문장 요약
- next_action: 반드시 "회신", "확인", "무시" 중 하나
- priority: 반드시 "높음", "보통", "낮음" 중 하나
- reason: 판단 근거
- suggested_reply: 고객에게 보낼 짧은 답변 초안

next_action 기준:
- "회신": 고객에게 바로 안내/사과/추가 정보 요청을 보내야 하는 경우
- "확인": 주문, 배송, 환불, 재고, 제품 상태, AS 가능 여부 등 내부 확인이 필요한 경우
- "무시": 구매 사실, 상품, 주문번호, 구체 요청이 없고 실제 처리할 내용이 없는 경우

priority 기준:
- "높음": 제품 파손, 안전/부작용, 환불 지연, 즉시 사용 불가 등 긴급성이 높은 경우
- "보통": 배송 지연, 구성품 누락, 일반 교환/반품/AS 문의
- "낮음": 처리할 수 없는 비구매성 비난 또는 정보가 거의 없는 경우
""",
)
````

````python
# 5) 분석 함수 작성
def analyze_one_complaint(complaint_id):
    # 1. Agent에게 보낼 질문 만들기
    user_message = f"complaint_id={complaint_id} 컴플레인을 조회해서 분석해줘."

    # 2. Agent 실행
    result = agent.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": user_message,
                }
            ]
        }
    )

    # 3. Agent의 구조화된 분석 결과 꺼내기
    structured_response = result["structured_response"]

    # 4. Pydantic 객체를 파이썬 딕셔너리로 변환
    analysis = structured_response.model_dump()

    # 5. 분석 결과에 complaint_id 추가
    analysis["complaint_id"] = complaint_id

    # 6. 최종 결과 반환
    return analysis
````

````python
# 6) CSV 분석

analysis_rows = []

for complaint_id in df["complaint_id"].tolist():
    print(f"분석 중: complaint_id={complaint_id}")
    analysis_rows.append(analyze_one_complaint(int(complaint_id)))

analysis_df = pd.DataFrame(analysis_rows)
analysis_df.head()
````

````python
# 7) 원본 데이터 일부와 분석 결과 합치기
final_df = df.merge(analysis_df, on="complaint_id", how="left")

final_columns = [
    "complaint_id",
    "customer_name",
    "email",
    "product_name",
    "purchase_channel",
    "order_id",
    "received_date",
    "complaint_text",
    "intent",
    "emotion",
    "priority",
    "next_action",
    "summary",
    "reason",
    "suggested_reply",
]

final_df = final_df[final_columns]
````

````python
# 8) 결과 CSV 저장
OUTPUT_PATH = "complaints_analyzed.csv"
final_df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8")

print(f"저장 완료: {OUTPUT_PATH}")
final_df.head()
````
